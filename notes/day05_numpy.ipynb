```python
#神经网络通常使用 float32，而且经常把像素缩放到 0～1
pixels = np.array([
    [0, 64, 128],
    [192, 224, 255]
], dtype=np.uint8)
normalized_pixels=pixels.astype(np.float32)/255.0
print(pixels)
print(pixels.dtype)
print(normalized_pixels)
print(normalized_pixels.dtype)

#reshape() 中的 -1
#如果只想指定一个维度，另一个维度让 NumPy 自动计算，可以使用 -1
numbers=np.arange(1,13)
a=numbers.reshape(3,-1)
b=numbers.reshape(-1,6)
print(a)
print(a.shape)
print(b)
print(b.shape)

#flatten() 与 ravel(),把多维数组变成一维
#flatten()：一定复制
matrix=np.arange(1,7).reshape(2,3)
flattened=matrix.flatten()
flattened[0]=100
print(flattened)
print(matrix)

#ravel()：尽量返回视图,修改 raveled 会影响 matrix,二者共享数据
matrix=np.arange(1,7).reshape(2,3)
raveled=matrix.ravel()
raveled[0]=100
print(raveled)
print(matrix)
```

```python
#完整对比实验
#np.shares_memory(a, b) 用来判断两个 NumPy 数组是否共享同一块底层内存
matrix = np.arange(1, 7).reshape(2, 3)

flattened = matrix.flatten()
raveled = matrix.ravel()

print("flatten 是否共享内存：",
      np.shares_memory(matrix, flattened))

print("ravel 是否共享内存：",
      np.shares_memory(matrix, raveled))
flattened[0] = 100
raveled[1] = 200

print("flattened：", flattened)
print("raveled：", raveled)
print("matrix：")
print(matrix)
```

```python
#聚合练习
scores = np.array([
    [80, 90, 70],
    [60, 75, 85],
    [95, 88, 92]
])

#所有成绩总和和总平均分
print(scores.sum())
print(scores.mean())

#每门课程平均分
print(scores.mean(axis=0))

#每名学生平均分
print(scores.mean(axis=1))

#每门课程最高分和最低分
print(scores.max(axis=0),scores.min(axis=0))

#每名学生最高分
print(scores.max(axis=1))

#每门课程最高分所在的学生下标
print(scores.argmax(axis=0))

#每名学生最高分所在的课程下标
print(scores.argmax(axis=1))

#每门课程标准差
print(scores.std(axis=0))

#使用 keepdims=True 计算每名学生平均分
print(scores.mean(axis=1,keepdims=True))

#打印所有结果的 shape，说明哪个维度被消掉了
course_mean = scores.mean(axis=0)
student_mean = scores.mean(axis=1)
student_mean_keepdims = scores.mean(axis=1, keepdims=True)

print("每门课程平均分 shape：", course_mean.shape)
print("每名学生平均分 shape：", student_mean.shape)
print("保留维度后的平均分 shape：", student_mean_keepdims.shape)

#让每名学生的成绩减去自己的平均分，打印结果
ans=scores-scores.mean(axis=1,keepdims=True)
print(ans)

#验证处理后每一行的平均值接近 0
print(ans.mean(axis=1))
```

```python
#对每门课程进行 Z-score 标准化
scores = np.array([
    [80, 90, 70],
    [60, 75, 85],
    [95, 88, 92]
], dtype=np.float32)

course_mean = scores.mean(axis=0, keepdims=True)
course_std = scores.std(axis=0, keepdims=True)

standardized_scores = (scores - course_mean) / course_std

print("每门课程平均分：")
print(course_mean)
print("shape:", course_mean.shape)

print("每门课程标准差：")
print(course_std)
print("shape:", course_std.shape)

print("标准化后的成绩：")
print(standardized_scores)

print("标准化后每列平均值：")
print(standardized_scores.mean(axis=0))

print("标准化后每列标准差：")
print(standardized_scores.std(axis=0))
```

```python
#Min-Max 缩放
course_min = scores.min(axis=0, keepdims=True)
course_max = scores.max(axis=0, keepdims=True)

scaled_scores = (
    (scores - course_min)
    / (course_max - course_min)
)

print("每列最小值：", course_min)
print("每列最大值：", course_max)

print("缩放后的成绩：")
print(scaled_scores)

print("缩放后每列最小值：")
print(scaled_scores.min(axis=0))

print("缩放后每列最大值：")
print(scaled_scores.max(axis=0))
```

```python
# 常数列与除零保护
data = np.array([
    [10, 1],
    [20, 1],
    [30, 1]
], dtype=np.float32)

mean = data.mean(axis=0, keepdims=True)
std = data.std(axis=0, keepdims=True)

#np.where(条件, 条件成立时取什么, 条件不成立时取什么)
safe_std = np.where(std == 0, 1, std)
standardized = (data - mean) / safe_std

print("平均值：", mean)
print("原标准差：", std)
print("安全标准差：", safe_std)
print("标准化结果：")
print(standardized)

print("处理后每列平均值：")
print(standardized.mean(axis=0))

print("处理后每列标准差：")
print(standardized.std(axis=0))

#Min-Max保护
minimum = data.min(axis=0, keepdims=True)
maximum = data.max(axis=0, keepdims=True)

data_range = maximum - minimum
safe_range = np.where(data_range == 0, 1, data_range)

scaled = (data - minimum) / safe_range

print("安全极差：", safe_range)
print("Min-Max 结果：")
print(scaled)
```

```python
#验证相同种子
rng1 = np.random.default_rng(seed=2026)
rng2 = np.random.default_rng(seed=2026)

data1 = rng1.normal(loc=75, scale=10, size=10)
data2 = rng2.normal(loc=75, scale=10, size=10)

print(data1)
print(data2)
print("两个结果是否完全相同：")
print(np.array_equal(data1, data2))

#如果种子不同
rng3 = np.random.default_rng(seed=2027)
data3 = rng3.normal(loc=75, scale=10, size=10)

print(np.array_equal(data1, data3))
```

```python
#完整统计练习
rng = np.random.default_rng(seed=2026)

course_means = np.array([78, 82, 75])
course_stds = np.array([10, 8, 12])

scores = rng.normal(        #正态分布
    loc=course_means,
    scale=course_stds,
    size=(100, 3)
)

scores = np.clip(scores, 0, 100)        #限制成绩的最值
scores = scores.astype(np.float32)

print("成绩矩阵形状：", scores.shape)
print("前 5 名学生：")
print(scores[:5])

print("每门课程平均分：")
print(scores.mean(axis=0))

print("每门课程标准差：")
print(scores.std(axis=0))

print("每门课程最低分：")
print(scores.min(axis=0))

print("每门课程最高分：")
print(scores.max(axis=0))

print("每门课程及格人数：")
print((scores >= 60).sum(axis=0))       #sum求和把True当作1累加

print("每门课程及格率：")
print((scores >= 60).mean(axis=0))

rng_again = np.random.default_rng(seed=2026)

scores_again = rng_again.normal(
    loc=course_means,
    scale=course_stds,
    size=(100, 3)
)

scores_again = np.clip(scores_again, 0, 100)
scores_again = scores_again.astype(np.float32)

print("相同种子能否复现：")
print(np.array_equal(scores, scores_again))
```
