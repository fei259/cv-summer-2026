```python
from pathlib import Path        #Path 用来处理文件和文件夹路径，比直接拼接字符串更方便

import numpy as np
import matplotlib.pyplot as plt     #主要用于画图

plt.rcParams["font.sans-serif"] = [     #设置 Matplotlib 绘图时使用的中文字体
    "Microsoft YaHei",
    "SimHei"
]

plt.rcParams["axes.unicode_minus"] = False      #设置坐标轴中的负号正常显示
```

```python
current_dir = Path.cwd()        #Path.cwd() 用来获取当前工作目录

if (current_dir / "notes").exists():
    project_root = current_dir
elif current_dir.name == "notes":
    project_root = current_dir.parent
else:
    raise RuntimeError("无法确定项目根目录")

output_dir = project_root / "results" / "week1"

output_dir.mkdir(
    parents=True,
    exist_ok=True
)

print("当前目录：", current_dir)
print("项目根目录：", project_root)
print("图片输出目录：", output_dir)
```

```python
#第一张折线图
test_numbers=np.arange(1,7)
scores=np.array([72, 78, 75, 83, 88, 91])
fig,ax=plt.subplots(figsize=(8,5))       #画布，坐标系

ax.plot(
    test_numbers,
    scores,
    color="royalblue",
    marker="o",
    linestyle="-",
    linewidth=2,
    label="测试成绩"
)

ax.set_title("六次测试成绩变化")
ax.set_xlabel("测试次数")
ax.set_ylabel("成绩")
ax.set_xticks(test_numbers)
ax.set_ylim(60,100)

ax.grid(
    True,
    linestyle="--",
    alpha=0.4       #透明度为 40%
)

ax.legend()     #显示图例(label="测试成绩")
fig.tight_layout()      #自动调整边距，避免标题和坐标轴文字被裁切
plt.show()      #在 Notebook 中显示图像
```

```python
test_numbers = np.arange(1, 7)

student_a = np.array([72, 78, 75, 83, 88, 91])
student_b = np.array([80, 79, 82, 81, 86, 89])

fig,ax=plt.subplots(figsize=(8,5))

ax.plot(
    test_numbers,
    student_a,
    color="royalblue",
    marker="o",
    linestyle="-",
    linewidth=2,
    label="学生 A"
)

ax.plot(
    test_numbers,
    student_b,
    color="orange",
    marker="s",
    linestyle="--",
    linewidth=2,
    label="学生 B"
)

ax.set_title("学生成绩对比")
ax.set_xlabel("测试次数")
ax.set_ylabel("成绩")
ax.set_xticks(test_numbers)
ax.set_ylim(60, 100)

ax.grid(
    True,
    linestyle="--",
    alpha=0.4
)

ax.legend()
fig.tight_layout()

fig.savefig(
    output_dir / "score_lines.png",
    dpi=150,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```

```python
subjects = np.array(["数学", "英语", "编程"])
average_scores = np.array([82, 85, 88])

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(
    subjects,
    average_scores,
    color="cornflowerblue",
    width=0.6,
    label="平均成绩"
)

ax.set_title("各科平均成绩")
ax.set_xlabel("课程")
ax.set_ylabel("平均分")

ax.set_ylim(0, 100)

ax.grid(
    axis="y",       #只显示 y 轴的网格线
    linestyle="--",
    alpha=0.4
)

ax.set_axisbelow(True)
ax.legend()

ax.bar_label(
    bars,       #在柱子上显示标签
    fmt="%.0f",     #保留0位小数
    padding=3       #标签与柱子顶部之间的距离
)

fig.tight_layout()
plt.show()
```

```python
#分组柱状练习
#ax.bar(x, height)
#ax.bar(subjects, average_scores)
# subjects：横坐标，也就是每根柱子放在哪里、叫什么
# average_scores：柱子的高度，也就是纵轴对应的数值
subjects = np.array(["数学", "英语", "编程"])

student_a = np.array([85, 90, 88])
student_b = np.array([78, 86, 92])

fig,ax=plt.subplots(figsize=(8,5))

x = np.arange(len(subjects))        #每门课程的中心位置
width = 0.35

bars_a = ax.bar(
    x - width / 2,      #学生 A 向左偏移半个柱子宽度
    student_a,
    width,
    label="学生 A"
)

bars_b = ax.bar(
    x + width / 2,      #学生 B 向右偏移半个柱子宽度
    student_b,
    width,
    label="学生 B"
)

ax.set_title("学生各科成绩对比")
ax.set_xlabel("课程")
ax.set_ylabel("成绩")
ax.set_xticks(x)
ax.set_xticklabels(subjects)
ax.set_ylim(0, 110)

ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.4
)

ax.set_axisbelow(True)
ax.legend()

ax.bar_label(
    bars_a,
    fmt="%.0f",
    padding=3
)

ax.bar_label(
    bars_b,
    fmt="%.0f",
    padding=3
)
    
fig.tight_layout()

fig.savefig(
    output_dir / "score_bars.png",
    dpi=150,
    bbox_inches="tight",
    facecolor="white"
)

plt.show() 
```

```python
math_scores = np.array([
    65, 70, 75, 78, 82,
    85, 88, 90, 93, 95
])

programming_scores = np.array([
    60, 68, 72, 76, 80,
    84, 89, 91, 94, 96
])

fig, ax = plt.subplots(figsize=(7, 6))

ax.scatter(
    math_scores,
    programming_scores,
    color="royalblue",
    s=70,       #点的面积大小
    alpha=0.8,      #点的透明度
    marker="o",
    edgecolors="black",     #点的边框颜色
    label="学生"
)

ax.set_title("数学成绩与编程成绩的关系")
ax.set_xlabel("数学成绩")
ax.set_ylabel("编程成绩")

ax.set_xlim(55, 100)
ax.set_ylim(55, 100)

ax.grid(
    True,
    linestyle="--",
    alpha=0.4
)

ax.legend()     #显示图例(label="学生")

fig.tight_layout()
plt.show()
```

```python
#散点图练习
math_scores = np.array([
    65, 70, 75, 78, 82,
    85, 88, 90, 93, 95
])

programming_scores = np.array([
    60, 68, 72, 76, 80,
    84, 89, 91, 94, 96
])

study_hours = np.array([
    2, 3, 4, 3.5, 5,
    5.5, 6, 7, 8, 9
])

fig,ax=plt.subplots(figsize=(7,6))

correlation=np.corrcoef(math_scores, programming_scores)[0, 1]      #计算数学成绩和编程成绩的相关系数

scatter=ax.scatter(
    math_scores,
    programming_scores,
    c=study_hours,     #点的颜色根据学习时间变化，使用学习时间控制散点颜色    
    cmap="viridis",    #使用 Viridis 颜色映射   
    s=80,
    alpha=0.5,
    marker="o",
    edgecolors="black",
    label="学生"
)

#添加标有“每周学习时长”的颜色条
colorbar=fig.colorbar(scatter, ax=ax)      #添加颜色条
colorbar.set_label("每周学习时长")

#添加灰色虚线 y=x
ax.plot(
    [55, 100],      #两个点的横坐标
    [55, 100],      #两个点的纵坐标
    color="gray",
    linestyle="--",
    linewidth=1,
    alpha=0.6,
    label="两科分数相同"
)

ax.set_title("数学成绩与编程成绩的关系")
ax.set_xlabel("数学成绩")
ax.set_ylabel("编程成绩")
ax.set_xlim(55, 100)
ax.set_ylim(55, 100)

#添加半透明虚线网格
ax.grid(
    True,
    linestyle="--",
    alpha=0.5
)

ax.legend()
fig.tight_layout()

fig.savefig(
    output_dir / "score_scatter.png",
    dpi=150,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()

#打印相关系数，保留三位小数
print(f"数学成绩与编程成绩的相关系数为：{correlation:.3f}")
```

```python
fig, axes = plt.subplots(
    1,
    3,
    figsize=(18, 5)
)       #创建一个包含 1 行 3 列子图的画布,axes[0]、axes[1]、axes[2] 分别对应三个子图

# 第一个子图：折线图
test_numbers = np.arange(1, 7)

student_a = np.array([72, 78, 75, 83, 88, 91])
student_b = np.array([80, 79, 82, 81, 86, 89])

axes[0].plot(
    test_numbers,
    student_a,
    marker="o",
    label="学生 A"
)

axes[0].plot(
    test_numbers,
    student_b,
    marker="s",
    linestyle="--",
    label="学生 B"
)

axes[0].set_title("测试成绩变化")
axes[0].set_xlabel("测试次数")
axes[0].set_ylabel("成绩")
axes[0].set_xticks(test_numbers)
axes[0].set_ylim(60, 100)
axes[0].grid(
    True,
    linestyle="--",
    alpha=0.4
)
axes[0].legend()

# 第二个子图：柱状图
subjects = np.array(["数学", "英语", "编程"])
average_scores = np.array([82, 85, 88])

bars = axes[1].bar(
    subjects,
    average_scores,
    color="cornflowerblue"
)

axes[1].set_title("各科平均成绩")
axes[1].set_xlabel("课程")
axes[1].set_ylabel("平均分")
axes[1].set_ylim(0, 100)
axes[1].grid(
    axis="y",
    linestyle="--",
    alpha=0.4
)
axes[1].set_axisbelow(True)
axes[1].bar_label(bars)

# 第三个子图：散点图
math_scores = np.array([
    65, 70, 75, 78, 82,
    85, 88, 90, 93, 95
])

programming_scores = np.array([
    60, 68, 72, 76, 80,
    84, 89, 91, 94, 96
])

study_hours = np.array([
    2, 3, 4, 3.5, 5,
    5.5, 6, 7, 8, 9
])

scatter = axes[2].scatter(
    math_scores,
    programming_scores,
    c=study_hours,
    cmap="viridis",
    s=70,
    alpha=0.8,
    edgecolors="black"
)

axes[2].set_title("数学与编程成绩")
axes[2].set_xlabel("数学成绩")
axes[2].set_ylabel("编程成绩")
axes[2].set_xlim(55, 100)
axes[2].set_ylim(55, 100)
axes[2].grid(
    True,
    linestyle="--",
    alpha=0.4
)

colorbar = fig.colorbar(
    scatter,
    ax=axes[2]
)
colorbar.set_label("每周学习时长")

# 整张画布的总标题
fig.suptitle(
    "第一周成绩数据可视化",
    fontsize=16
)

fig.tight_layout(
    rect=(0, 0, 1, 0.94)
)

fig.savefig(
    output_dir / "visualization_summary.png",
    dpi=150,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```
